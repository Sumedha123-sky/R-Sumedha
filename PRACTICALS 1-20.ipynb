```python
# Conditional Statements and Loops

# 1. Check Positive Number (if)
num = int(input("Enter a number: "))
if num > 0:
    print("Positive Number")

# 2. Even or Odd (if-else)
num = int(input("Enter another number: "))
if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")

# 3. Grade (if-elif-else)
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Fail")

# 4. Login (Nested if-else)
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Incorrect Password")
else:
    print("Invalid Username")
    # 5. Voting Eligibility (One-line if-else)
age = int(input("Enter your age: "))
print("Eligible to Vote" if age >= 18 else "Not Eligible to Vote")


# 6. For Loop
print("Numbers using for loop:")
for i in range(1, 6):
    print(i)

# 7. While Loop
print("Numbers using while loop:")
i = 1
while i <= 5:
    print(i)
    i += 1

# 8. Break Statement
print("Break Example:")
for i in range(1, 11):
    if i == 5:
        break
        print(i)

# 9. Continue Statement
print("Continue Example:")
for i in range(1, 11):
    if i == 5:
        continue
    print(i)

# 10. Pass Statement
print("Pass Example:")
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
# 11. Match-Case Statement
day = int(input("Enter day number (1-7): "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Day")

print("Program Ended")
```

# PRACTICAL 3

```python
# 1. No arguments, No return value
def greet():
    print("Hello!")
greet()
# 2. Arguments, No return value
def add(a, b):
    print("Sum =", a + b)

add(10, 20)

# 3. No arguments, Return value
def message():
    return "Welcome to Python"

print(message())

# 4. Arguments, Return value
def multiply(a, b):
    return a * b

print("Product =", multiply(5, 4))

# 5. Default argument
def student(name="Srija"):
    print("Name:", name)

student()
student("Ravi")

# 6. Keyword arguments
def details(name, age):
    print("Name:", name)
    print("Age:", age)

details(age=18, name="Anu")

# 7. Variable length arguments
def total(*marks):
    print("Total Marks =", sum(marks))
total(80, 90, 85)

# 8. Lambda function
square = lambda x: x * x
print("Square =", square(5))

# 9. Recursive function
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print("Factorial =", fact(5))

# 10. Nested function
def outer():
    def inner():
        print("Inner Function")
    inner()

outer()
```

```python
import pandas as pd
import numpy as np

# 1. Create DataFrame containing missing values
data = {
    'Name': ['Arun', 'Bala', 'Cathy', 'Divya', None],
    'Age': [20, np.nan, 19, 22, 21],
    'Marks': [85, 90, np.nan, 88, np.nan]
}

df = pd.DataFrame(data)

# 2. Display original DataFrame
print("Original DataFrame:")
print(df)

# 3. Check missing values using isnull()
print("\nMissing Values:")
print(df.isnull())

# 4. Count missing values in each column
print("\nNumber of Missing Values:")
print(df.isnull().sum())

# 5. Fill all missing values with a default value
df_default = df.fillna("Not Available")


print("\nAfter Filling Missing Values with Default Value:")
print(df_default)

# 6. Fill missing Marks values with mean
df_mean = df.copy()
df_mean['Marks'] = df_mean['Marks'].fillna(df_mean['Marks'].mean())

print("\nAfter Filling Missing Marks with Mean:")
print(df_mean)

# 7. Forward fill missing values
df_ffill = df.ffill()

print("\nAfter Forward Fill:")
print(df_ffill)

# 8. Backward fill missing values
df_bfill = df.bfill()

print("\nAfter Backward Fill:")
print(df_bfill)

# 9. Remove rows containing missing values
df_drop = df.dropna()

print("\nAfter Removing Missing Value Rows:")
print(df_drop)

# 10. Replace a specified value
df_replace = df.fillna("Missing")
df_replace = df_replace.replace("Missing", "Not Available")

print("\nAfter Replacing Value:")
print(df_replace)

# 11. Display DataFrame information
print("\nDataFrame Information:")
df.info()
```

```python
import pandas as pd

# 1. Create DataFrame containing employee details
data = {
    'Name': ['Arun', 'Bala', 'Cathy', 'Divya', 'Esha', 'Farhan'],
    'Department': ['IT', 'HR', 'IT', 'HR', 'Finance', 'IT'],
    'Experience': [2, 5, 3, 4, 6, 2],
    'Salary': [30000, 40000, 35000, 45000, 50000, 32000]
}

df = pd.DataFrame(data)

# 2. Display original DataFrame
print("Original DataFrame:")
print(df)

# 3. Group data based on Department
grouped = df.groupby('Department')

print("\nGrouped Data:")
print(grouped.groups)

# 4. Total salary for each department
print("\nTotal Salary by Department:")
print(df.groupby('Department')['Salary'].sum())

# 5. Average salary for each department
print("\nAverage Salary by Department:")
print(df.groupby('Department')['Salary'].mean())

# 6. Maximum and minimum salary
print("\nMaximum Salary by Department:")
print(df.groupby('Department')['Salary'].max())

print("\nMinimum Salary by Department:")
print(df.groupby('Department')['Salary'].min())

# 7. Count employees in each department
print("\nNumber of Employees in Each Department:")
print(df.groupby('Department')['Name'].count())

# 8. Multiple aggregation operations
print("\nMultiple Aggregation Operations:")
print(df.groupby('Department')['Salary'].agg(['sum', 'mean', 'max', 'min']))

# 9. Group using Department and Experience
print("\nGrouping by Department and Experience:")
print(df.groupby(['Department', 'Experience'])['Salary'].sum())

# 10. Iterate through each group
print("\nIterating Through Groups:")

for department, group in df.groupby('Department'):
    print("\nDepartment:", department)
    print(group)

# 11. Sort DataFrame by Salary
print("\nData Sorted by Salary:")
print(df.sort_values(by='Salary'))

# 12. End of Program
print("\nProgram Completed Successfully!")
```

```python
# Merge and Join DataFrames using Pandas

import pandas as pd

# ----- CONCATENATE OPERATIONS ------------------- (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 1:")
print(df1)

#  Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])   

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```

```python
# Basic Data Cleaning and Preprocessing

import pandas as pd
import csv

# Loading csv file 
file_path =r"C:\Users\Sumedha\Desktop\Weekly Task\students.csv"

# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

# 3. Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# 4. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 5. Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# 7. Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# 8. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 9. Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# 10. Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# 11. Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# 12. Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# 13. Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# 14. Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# 15. Reset the index
df = df.reset_index(drop=True)

# 16. Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# 17. Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# 18. Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
import pandas as pd
import numpy as np

# Create sample dataset
data = {
    'Name': ['Arun', 'Bala', 'Charan', 'Divya', 'Bala'],
    'Age': [20, 21, np.nan, 23, 21],
    'Marks': [90, 85, 78, np.nan, 85],
    'Gender': ['Male', 'Male', 'Male', np.nan, 'Male']
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing numerical values with mean
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Fill missing categorical values
df['Gender'] = df['Gender'].fillna('Unknown')

# Remove duplicate records
df = df.drop_duplicates()

# Convert Age to integer
df['Age'] = df['Age'].astype(int)

# Rename Marks column to Score
df = df.rename(columns={'Marks': 'Score'})

# Sort by Score in descending order
df = df.sort_values(by='Score', ascending=False)

# Display records where Score > 85
print("\nRecords with Score greater than 85:")
print(df[df['Score'] > 85])

# Display final dataset
print("\nFinal Dataset:")
print(df)

# Display dataset information
print("\nDataset Information:")
df.info()
```
