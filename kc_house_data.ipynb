```python
# Number of rows before removing null values
print("Rows before removing null values:", df.shape[0])

# Remove rows containing null values
df = df.dropna()

# Number of rows after b removing null values
print("Rows after removing null values:", df.shape[0])
```

```python
# statistical summary
df.describe(include="all")
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_summary
```

```python
print("Total missing values:", df.isnull().sum())
```

```python
# Numerical columns
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Categorical columns
categorical_cols = df.select_dtypes(include=["object", "str"]).columns

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```
