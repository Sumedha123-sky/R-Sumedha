```python
print("Dataset Shape:", df.shape)
print("\nDataset Information:")
print(df.info())

print("\nColumn Names:")
print(df.columns)

print("\nStatistical Description:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Values:")
print(df.duplicated().sum())
```

```python
print("EDA COMPLETED SUCCESSFULLY")

print("Total Students:", len(df))
print("Average Marks:", df['Marks'].mean())
print("Highest Marks:", df['Marks'].max())
print("Lowest Marks:", df['Marks'].min())

print("\nDepartment Count:")
print(df['Dept'].value_counts())
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Dept')
plt.title('Department-wise Student Count')
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['Marks'], bins=5, edgecolor='black')
plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

```python
dept_marks = df.groupby('Dept')['Marks'].mean()

plt.figure(figsize=(6,4))
dept_marks.plot(kind='bar')
plt.title('Average Marks by Department')
plt.xlabel('Department')
plt.ylabel('Average Marks')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(y=df['Marks'])
plt.title('Student Marks Boxplot')
plt.ylabel('Marks')
plt.show()
```

```python
dept_count = df['Dept'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(dept_count, labels=dept_count.index,
        autopct='%1.1f%%', startangle=90)
plt.title('Department Distribution')
plt.show()
```
